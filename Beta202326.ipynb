```python
import yfinance as yf
import pandas as pd
import numpy as np
```

```python

# Define the four stocks
stocks = ["AAPL", "JPM", "XOM", "KO"]

# Define the market benchmark
market = "^GSPC"

# Combine all tickers
tickers = stocks + [market]

# Define the sample period
start_date = "2023-01-01"
end_date = "2026-01-01"

# Download historical daily data
data = yf.download(
    tickers,
    start=start_date,
    end=end_date,
    auto_adjust=True,
    progress=False
)

# Extract adjusted closing prices
prices = data["Close"]

# Display first five observations
print(prices.head(12))
```

```python

# Calculate daily percentage returns
returns = prices.pct_change(fill_method=None)

# Remove rows containing missing values
returns = returns.dropna()

# Display first five rows
print(returns.head())
```

```python

# Extract S&P 500 daily returns
market_returns = returns["^GSPC"]

# Calculate market variance
market_variance = market_returns.var()

print("Market Variance:", market_variance)
```

```python

# Extract Apple daily returns
apple_returns = returns["AAPL"]

# Calculate covariance
apple_covariance = apple_returns.cov(market_returns)

# Calculate beta
apple_beta = apple_covariance / market_variance

print("Apple Beta:", round(apple_beta, 4))
```

```python

# Create an empty dictionary
betas = {}

# Calculate beta for each stock
for stock in stocks:

    stock_returns = returns[stock]

    covariance = stock_returns.cov(market_returns)

    beta = covariance / market_variance

    betas[stock] = beta

# Convert results into a DataFrame
beta_df = pd.DataFrame(
    list(betas.items()),
    columns=["Stock", "Beta"]
)

# Round beta values
beta_df["Beta"] = beta_df["Beta"].round(4)

print(beta_df)
```
